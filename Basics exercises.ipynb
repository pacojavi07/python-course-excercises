### Basics exercises

1. Find the minimum number

```python
def minimum(a, b):
    if a <= b:
        return a
    else:
        return b
    
a = 2
b = 4
print(minimum(a, b))
```

2. Reversing words of a string

```python
def rev_words(sentence):
    
    words = sentence.split()
    
    reverse_sent = ' '.join(reversed(words))
    
    return reverse_sent

rev_words("string de prueba")
```

3. Suming up elements of a tuple

```python
tup = (1, 2, 3, 4)

def sum_tuple(tupla):
    
    return sum(list(tupla))

sum_tuple(tup)
```

6. Check if a number is prime

```python
def isprime(num):
    if num < 1:
        return False
    elif num == 2:
        return True
    else:
        for i in range (2, num):
            if num % i == 0:
                return False
        return True
    
print(isprime(5))    
```
